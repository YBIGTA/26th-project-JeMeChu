```python
import pandas as pd

df = pd.read_csv("final_sql.csv",encoding="utf-8-sig")
url = pd.read_csv("link_url.csv",encoding="utf-8-sig")
```

```python
df.head(1)
url.head(1)
```

```python
url = url.rename(columns={'img_urls': 'connect_urls'})
```

```python
url.head(1)
url = url[["id","connect_urls"]]
```
