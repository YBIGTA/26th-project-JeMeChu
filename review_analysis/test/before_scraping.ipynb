```python
df = pd.read_csv("photo_url.csv",encoding="utf-8-sig")
```

```python
df['photo_url'] = df['photo_url'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[") else x)
```

```python
# 길이를 안전하게 계산하는 함수 정의
def safe_len(value):
    if isinstance(value, float):  # float이면 0 반환 (NaN 포함)
        return 0
    return len(value)  # 문자열 변환 후 길이 반환

# 길이가 3이 아닌 행 필터링
df_filtered = df[df['photo_url'].apply(safe_len) != 3]

# 결과 출력
print(df_filtered)
```
