```python
import ast
from collections import Counter
```

```python
def get_unique_features_with_counts(df):
    unique_values = {}

    for column in df.columns:
        all_values = []  # Use a list to track occurrences

        for row in df[column].dropna():
            # Convert string representation of list to an actual list if necessary
            if isinstance(row, str):
                try:
                    row = ast.literal_eval(row)
                except (SyntaxError, ValueError):
                    pass  # If it fails, it's a regular string and should be treated as such

            # If it's a list, unpack it; otherwise, add directly
            if isinstance(row, list):
                for item in row:
                    if isinstance(item, list) or isinstance(item, tuple):
                        all_values.extend(item)  # Unpack nested structures
                    else:
                        all_values.append(item)
            else:
                all_values.append(row)  # Directly add non-list values

        # Remove numeric values and count occurrences
        text_only_values = [v for v in all_values if isinstance(v, str)]
        counts = Counter(text_only_values)

        unique_values[column] = dict(counts)

    return unique_values
```

```python
# Get unique features
df = df[["facilities","parking","very_good","seat_info"]]
get_unique_features_with_counts = get_unique_features_with_counts(df)
```

```python
file_path = "unique_features.txt"
with open(file_path, "w", encoding="utf-8") as f:
    for column, values in get_unique_features_with_counts.items():
        f.write(f"{column}:\n")
        for value, count in values.items():
            f.write(f"- {value}: {count}\n")
        f.write("\n")  # Newline for readability
```
