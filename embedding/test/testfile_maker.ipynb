```python
import pandas as pd

df = pd.read_csv("pre_embedding.csv", encoding = "utf-8-sig")
```

```python
df.columns
```

```python
df = df[["id","description","latest_reviews"]]
```

```python
df["latest_reviews"] = df["latest_reviews"].apply(lambda x: eval(x) if isinstance(x, str) else x)  # Convert string lists to actual lists
latest_df = df[["id","latest_reviews"]].explode("latest_reviews")
description_df = df[["id","description"]]
```

```python
description_df
```

```python

latest_df.rename(columns={"latest_reviews": "embedded"}, inplace=True)
description_df.rename(columns={"description": "embedded"}, inplace=True)
```

```python
df_combined = pd.concat([latest_df, description_df], ignore_index=True)
```

```python
df_combined
```

```python
df_combined.to_csv("pre_embedding_ready.csv",encoding="utf-8-sig",index=False)
```

```python
total = total.merge(drop[["embedded", "embedding"]], on="embedded", how="left")
```

```python
# Merge and directly update the 'embedding' column
total = total.merge(drop[["embedded", "embedding"]], on="embedded", how="left", suffixes=("", "_drop"))

# Fill missing values in 'embedding' with values from 'embedding_drop'
total["embedding"] = total["embedding"].combine_first(total["embedding_drop"])

# Drop the extra column
total.drop(columns=["embedding_drop"], inplace=True)
```
